#### Production Caching Architecture

Our LLMOps library implements sophisticated caching at multiple levels:

**Embedding Caching:**
The process of embedding is typically very time consuming and expensive:

1. Send text to OpenAI API endpoint
2. Wait for processing  
3. Receive response
4. Pay for API call

This occurs *every single time* a document gets converted into a vector representation.

**Our Caching Solution:**
1. Check local cache for previously computed embeddings
2. If found: Return cached vector (instant, free)
3. If not found: Call OpenAI API, store result in cache
4. Return vector representation

**LLM Response Caching:**
Similarly, we cache LLM responses to avoid redundant API calls for identical prompts.

**Benefits:**
- ⚡ Faster response times (cache hits are instant)
- 💰 Reduced API costs (no duplicate calls)  
- 🔄 Consistent results for identical inputs
- 📈 Better scalability

Our ProductionRAGChain automatically handles all this caching behind the scenes!

````python
# Let's test our Production RAG Chain to see caching in action
print("Testing RAG Chain with caching...")

# Test query
test_question = "What is this document about?"

try:
    # First call - will hit OpenAI API and cache results
    print("\n🔄 First call (cache miss - will call OpenAI API):")
    import time
    start_time = time.time()
    response1 = rag_chain.invoke(test_question)
    first_call_time = time.time() - start_time
    print(f"Response: {response1.content[:200]}...")
    print(f"⏱️ Time taken: {first_call_time:.2f} seconds")
    
    # Second call - should use cached results (much faster)
    print("\n⚡ Second call (cache hit - instant response):")
    start_time = time.time()
    response2 = rag_chain.invoke(test_question)
    second_call_time = time.time() - start_time
    print(f"Response: {response2.content[:200]}...")
    print(f"⏱️ Time taken: {second_call_time:.2f} seconds")
    
    speedup = first_call_time / second_call_time if second_call_time > 0 else float('inf')
    print(f"\n🚀 Cache speedup: {speedup:.1f}x faster!")
    
    # Get retriever for later use
    retriever = rag_chain.get_retriever()
    print("✓ Retriever extracted for agent integration")
    
except Exception as e:
    print(f"❌ Error testing RAG chain: {e}")
    retriever = None
````

##### ❓ Question #1: Production Caching Analysis

What are some limitations you can see with this caching approach? When is this most/least useful for production systems? 

Consider:
- **Memory vs Disk caching trade-offs**
- **Cache invalidation strategies** 
- **Concurrent access patterns**
- **Cache size management**
- **Cold start scenarios**

> NOTE: There is no single correct answer here! Discuss the trade-offs with your group.

##### ✅ Answer

1. Memory vs disk: Memory is fast but limited; disk is slower but persistent.

2. Cache invalidation: Hard to know when to clear/update stale data.

3. Concurrent access: Multiple users can cause race conditions or contention.

4. Cache size: Must manage what stays (LRU/eviction policies).

5. Cold start: First requests are slow since there’s no cache.

- **What caching strategies work best for each agent type?**

````python
# Activity 3: Building a Production-Safe LangGraph Agent with Guardrails

from langgraph_agent_lib.guardrails import (
    create_guardrails_guard,
    create_factuality_guard,
    validate_input,
    validate_output,
    GuardrailsState
)
from langgraph_agent_lib.agents import create_langgraph_agent, AgentState
from langgraph.graph import StateGraph, END
from langchain_core.messages import HumanMessage, AIMessage
from guardrails.hub import ProfanityFree
from guardrails import Guard
import logging

# Set up logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Create input guard for pre-processing
print("Setting up Input Guardrails...")
input_guard = create_guardrails_guard(
    valid_topics=["student loans", "financial aid", "education financing", "loan repayment"],
    invalid_topics=["investment advice", "crypto", "gambling", "politics"],
    enable_jailbreak_detection=True,
    enable_pii_protection=True,
    enable_profanity_check=True
)
print("✓ Input guard configured (jailbreak, topic, PII, profanity)")

# Create output guard for post-processing
print("\nSetting up Output Guardrails...")
output_guard = Guard().use(
    ProfanityFree(threshold=0.8, validation_method="sentence", on_fail="exception")
)
print("✓ Output guard configured (content moderation)")

# Create factuality guard (optional, for RAG responses)
print("\nSetting up Factuality Guard...")
factuality_guard = create_factuality_guard(
    eval_model="gpt-4.1-mini",
    on_prompt=True
)
print("✓ Factuality guard configured")
````

````python
# Define extended state for guarded agent
from typing_extensions import TypedDict, Annotated
from langgraph.graph.message import add_messages

class GuardedAgentState(TypedDict):
    """Extended state for guarded agent."""
    messages: Annotated[list, add_messages]
    validation_results: list
    should_reject: bool
    rejection_reason: str
    refinement_count: int

# Input validation node - handles jailbreak, topic, PII detection
def validate_input_node(state: GuardedAgentState) -> dict:
    """Validate user input before processing.
    
    Implements:
    - Jailbreak detection
    - Topic restriction
    - PII detection
    - Profanity check
    """
    messages = state.get("messages", [])
    if not messages:
        return {"should_reject": False, "validation_results": []}
    
    last_message = messages[-1]
    
    if isinstance(last_message, HumanMessage):
        try:
            user_input = last_message.content
            logger.info(f"🔍 Validating input: {user_input[:100]}...")
            
            # Validate input with guardrails
            result = validate_input(
                input_guard,
                user_input,
                raise_on_failure=False  # Don't raise, handle gracefully
            )
            
            validation_result = {
                "type": "input",
                "passed": result["validation_passed"],
                "timestamp": time.time(),
                "message": user_input[:100]
            }
            
            if not result["validation_passed"]:
                error_msg = result.get("error", "Input validation failed")
                logger.warning(f"❌ Input validation failed: {error_msg}")
                
                # Determine rejection reason
                if "jailbreak" in str(error_msg).lower():
                    rejection_reason = "This request appears to be attempting to bypass safety measures."
                elif "topic" in str(error_msg).lower() or "invalid topics" in str(error_msg).lower():
                    rejection_reason = "This question is outside my area of expertise. I can only help with student loans, financial aid, and education financing."
                elif "pii" in str(error_msg).lower():
                    rejection_reason = "I cannot process requests containing sensitive personal information. Please rephrase your question without including personal details."
                elif "profanity" in str(error_msg).lower():
                    rejection_reason = "I cannot process requests containing inappropriate language. Please rephrase your question."
                else:
                    rejection_reason = "This request does not meet our safety requirements."
                
                validation_result["error"] = error_msg
                return {
                    "should_reject": True,
                    "rejection_reason": rejection_reason,
                    "validation_results": [validation_result]
                }
            else:
                logger.info("✅ Input validation passed")
                return {
                    "should_reject": False,
                    "validation_results": [validation_result]
                }
                
        except Exception as e:
            logger.error(f"❌ Input validation error: {e}", exc_info=True)
            return {
                "should_reject": True,
                "rejection_reason": f"An error occurred during validation: {str(e)}",
                "validation_results": [{
                    "type": "input",
                    "passed": False,
                    "error": str(e)
                }]
            }
    
    return {"should_reject": False, "validation_results": []}

# Output validation node - handles content moderation and factuality
def validate_output_node(state: GuardedAgentState) -> dict:
    """Validate agent output after processing.
    
    Implements:
    - Content moderation (profanity)
    - Factuality checking (if context available)
    """
    messages = state.get("messages", [])
    if not messages:
        return {"validation_results": []}
    
    last_message = messages[-1]
    
    if isinstance(last_message, AIMessage):
        try:
            agent_output = last_message.content
            logger.info(f"🔍 Validating output: {agent_output[:100]}...")
            
            validation_results = []
            
            # Validate with content moderation guard
            if output_guard:
                result = validate_output(
                    output_guard,
                    agent_output,
                    raise_on_failure=False
                )
                
                output_result = {
                    "type": "output",
                    "passed": result["validation_passed"],
                    "guard": "content_moderation",
                    "timestamp": time.time()
                }
                
                if not result["validation_passed"]:
                    error_msg = result.get("error", "Output validation failed")
                    logger.warning(f"❌ Output validation failed: {error_msg}")
                    output_result["error"] = error_msg
                    
                    # Optionally sanitize the response
                    if result.get("validated_output"):
                        sanitized_message = AIMessage(
                            content=result["validated_output"]
                        )
                        validation_results.append(output_result)
                        return {
                            "messages": [sanitized_message],
                            "validation_results": validation_results
                        }
                else:
                    logger.info("✅ Output validation passed")
                
                validation_results.append(output_result)
            
            # Validate with factuality guard (if RAG context available)
            # This would check if the response aligns with retrieved documents
            # For now, we'll skip this or make it optional
            
            return {"validation_results": validation_results}
            
        except Exception as e:
            logger.error(f"❌ Output validation error: {e}", exc_info=True)
            return {
                "validation_results": [{
                    "type": "output",
                    "passed": False,
                    "error": str(e)
                }]
            }
    
    return {"validation_results": []}

# Rejection response node - handles guard failures gracefully
def rejection_node(state: GuardedAgentState) -> dict:
    """Generate a helpful rejection response when validation fails."""
    from langchain_core.messages import AIMessage
    
    rejection_reason = state.get("rejection_reason", 
                                 "This request does not meet our safety requirements.")
    
    rejection_message = AIMessage(
        content=f"I'm sorry, but I cannot process that request. {rejection_reason}\n\n"
                f"I can help you with questions about:\n"
                f"- Student loans and repayment options\n"
                f"- Financial aid programs\n"
                f"- Education financing\n"
                f"- Loan forgiveness programs\n\n"
                f"Please rephrase your question to focus on these topics."
    )
    
    logger.info("🚫 Request rejected - providing helpful error message")
    
    return {
        "messages": [rejection_message],
        "should_reject": False  # Reset for next request
    }

# Routing function
def should_reject(state: GuardedAgentState) -> str:
    """Route based on validation result."""
    if state.get("should_reject", False):
        return "reject"
    return "process"
````

````python
import time

def create_guarded_agent(
    rag_chain,
    model_name: str = "gpt-4.1-mini",
    temperature: float = 0.1
):
    """Create a LangGraph agent with integrated Guardrails."""
    
    # Create the base agent
    base_agent = create_langgraph_agent(
        model_name=model_name,
        temperature=temperature,
        rag_chain=rag_chain
    )
    
    # Build the guarded agent graph
    guarded_graph = StateGraph(GuardedAgentState)
    
    # Add nodes
    guarded_graph.add_node("validate_input", validate_input_node)
    guarded_graph.add_node("agent", lambda state: base_agent.invoke(state))
    guarded_graph.add_node("validate_output", validate_output_node)
    guarded_graph.add_node("reject", rejection_node)
    
    # Set entry point
    guarded_graph.set_entry_point("validate_input")
    
    # Add conditional edges
    guarded_graph.add_conditional_edges(
        "validate_input",
        should_reject,
        {
            "reject": "reject",
            "process": "agent"
        }
    )
    
    # Add sequential edges
    guarded_graph.add_edge("agent", "validate_output")
    guarded_graph.add_edge("validate_output", END)
    guarded_graph.add_edge("reject", END)
    
    return guarded_graph.compile()

# Create the guarded agent
print("\n" + "=" * 60)
print("Creating Guarded Agent...")
print("=" * 60)

guarded_agent = create_guarded_agent(
    rag_chain=rag_chain,
    model_name="gpt-4.1-mini",
    temperature=0.1
)

print("✓ Guarded Agent created successfully!")
print("\n🛡️ Security Features:")
print("  1. ✅ Input validation (jailbreak, topic, PII, profanity)")
print("  2. ✅ Output validation (content moderation)")
print("  3. ✅ Graceful error handling")
print("  4. ✅ Helpful rejection messages")
````
